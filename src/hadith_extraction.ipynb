```python
import json
import re

# ---------- تنظيف النص ----------
TASHKEEL = "".join(chr(c) for c in list(range(0x064B, 0x0660)) + [0x0670])


def strip_tashkeel(s):
    return s.translate(str.maketrans("", "", TASHKEEL))


def clean_text(s):
    s = strip_tashkeel(s)
    s = re.sub(r"[\(\[]\s*[0-9٠-٩]+", " ", s)
    for ch in "\u00ab\u00bb\"\u201c\u201d\u2018\u2019()[]{}":
        s = s.replace(ch, " ")
    s = s.replace("٠", " ").replace("؛", " ")
    s = re.sub(r"\bالله الله\b", "الله", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s


def clean_narrator(s):
    s = clean_text(s)
    s = s.replace(" - ", " ").replace(" _ ", " ")
    s = s.strip(" -_–—")
    s = re.sub(r"\s+", " ", s).strip()
    return s

# ---------- تبسيط التخريج ----------
IMAMS = [
    ("البخاري", "البخاري"),
    ("مسلم", "مسلم"),
    ("أبو داود", "أبو داود"),
    ("الترمذي", "الترمذي"),
    ("النسائي", "النسائي"),
    ("ابن ماجه", "ابن ماجه"),
    ("ابن ماجة", "ابن ماجه"),
    ("أحمد", "أحمد"),
    ("الموطأ", "مالك"),
    ("مالك", "مالك"),
    ("الدارمي", "الدارمي"),
    ("الدارقطني", "الدارقطني"),
    ("الحاكم", "الحاكم"),
    ("البيهقي", "البيهقي"),
    ("الطبراني", "الطبراني"),
]


def simplify_source(raw):
    found = []
    for key, name in IMAMS:
        if key in raw and name not in found:
            found.append(name)
    if found:
        return "رواه " + " و".join(found)
    return clean_text(raw)

# ---------- اكتشاف بدايات الأحاديث ----------
AN_RE = re.compile(r"^\s*[-_–—ـ]?\s*[0-9٠-٩]*\s*[ذ\-_–—ـ]?\s*عن\s+")
FOOT_RE = re.compile(r"^\s*[\(\[]\s*[0-9٠-٩]+[\)\]]")
TAKHRIJ_RE = re.compile(r"\b(?:رواه|البخاري|مسلم|الترمذي|أبو داود|النسائي|ابن ماجه|أحمد|الدارمي|الدارقطني|البيهقي|حديث حسن|حديث صحيح|أخرجه|متفق عليه)\b")
NARR_RE = re.compile(r"عن\s+(.+?رضي الله.*?(?:عنهما|عنهم|عنها|عنه))")
NARR_FALLBACK_RE = re.compile(r"عن\s+(.+?)\s+قال\b")


def is_headery(line):
    s = line.strip()
    if not s:
        return False
    if AN_RE.match(s):
        return False
    if FOOT_RE.match(s):
        return False
    if s.startswith("قوله"):
        return False
    if s[0] in "()[]{}":
        return False
    if TAKHRIJ_RE.search(s):
        return False
    if len(s.split()) > 9:
        return False
    return True


def preview_clean(ln):
    s = ln.strip()
    if not s or FOOT_RE.match(s):
        return None
    if s.startswith("بسم الله الرحمن الرحيم") and len(s) > 30:
        s = "الرحمن" + s[22:]
    return s


def valid_narrator_at(flat, k):
    plines = []
    for _, _, ln in flat[k:k + 8]:
        if AN_RE.match(ln) and plines:
            break
        if "رواه" in ln and plines:
            break
        c = preview_clean(ln)
        if c:
            plines.append(c)
        if len(plines) >= 6:
            break
    preview = " ".join(plines)
    m = NARR_RE.search(preview)
    if not m:
        m = NARR_FALLBACK_RE.search(preview)
    return bool(m) and len(m.group(1).split()) <= 18


flat = []
for pi, p in enumerate(pages):
    for li, ln in enumerate(p["text"].split("\n")):
        flat.append((pi, li, ln))

# سطر الراوي الحقيقي: سطر عن يسبقه عنوان + يليه راو صحيح
an_lines = []
for k, (pi, li, ln) in enumerate(flat):
    if not AN_RE.match(ln):
        continue
    lines = pages[pi]["text"].split("\n")
    j = li - 1
    while j >= 0 and (not lines[j].strip() or is_headery(lines[j])):
        j -= 1
    if j + 1 < li and valid_narrator_at(flat, k):
        an_lines.append((pi, j + 1))

assert len(an_lines) == 42, f"found {len(an_lines)} hadith starts"

HEADER_SET = set()
for _pi, _li in an_lines:
    _lns = pages[_pi]["text"].split("\n")
    _j = _li - 1
    while _j >= 0 and (not _lns[_j].strip() or is_headery(_lns[_j])):
        if _lns[_j].strip():
            HEADER_SET.add(_lns[_j].strip())
        _j -= 1

# ---------- تقسيم الراوي / المتن / التخريج ----------
PREAMBLE_RE = re.compile(
    r"^\s*((أن|عن)\s+)?(سمعت|سمعنا|قال|قالت|حدثني|حدثنا|أخبرني|أخبرنا|حفظت)?\s*(أيضا|أيضًا|ايضا)?\s*(رسول الله صلى الله عليه وسلم|النبي صلى الله عليه وسلم)?\s*(يقول|قال)?\s*[:٠\s«»\"\(\)\[\]\{\}\.،؛\-–—_]*")
SRC_STOP_RE = re.compile(r"[\(\[]\s*[0-9٠-٩]|قوله|أخرجه")
FOOTLINE_RE = re.compile(r"^\s*[\(\[]\s*[0-9٠-٩]+[\)\]]")
HADITH_GRADE_RE = re.compile(r"(حديث (?:حسن|صحيح|ضعيف).*)$")


RUNNING_RE = re.compile(r"^\s*[0-9٠-٩]*\s*ال[أآا]?ربعين النووية\s*[0-9٠-٩]*\s*$")


def parse_span(lines):
    fixed = []
    for ln in lines:
        s = ln.strip()
        if FOOTLINE_RE.match(s) or s.startswith("قوله"):
            continue
        if RUNNING_RE.match(s):
            continue
        if s in HEADER_SET:
            continue
        if s.startswith("بسم الله الرحمن الرحيم") and len(s) > 30:
            s = "الرحمن" + s[22:]
        fixed.append(s)
    joined = " ".join(t for t in fixed if t)

    m = NARR_RE.search(joined)
    if not m:
        m = NARR_FALLBACK_RE.search(joined)
    assert m, "no narrator in span: " + joined[:80]
    assert len(m.group(1).split()) <= 18, "narrator too long: " + m.group(1)[:80]
    narrator = clean_narrator(m.group(1))
    rest = joined[m.end():]
    for _ in range(3):
        new_rest = PREAMBLE_RE.sub("", rest, count=1)
        if new_rest == rest:
            break
        rest = new_rest

    ri = rest.find("رواه")
    if ri >= 0:
        matn_raw, src_raw = rest[:ri], rest[ri:]
        sm = SRC_STOP_RE.search(src_raw)
        if sm:
            src_raw = src_raw[:sm.start()]
        source = simplify_source(src_raw)
    else:
        gm = HADITH_GRADE_RE.search(rest)
        assert gm, "no source in span: " + rest[:80]
        source = simplify_source(gm.group(1))
        matn_raw = rest[:gm.start()]

    matn_raw = re.split(r"[\(\[]?\s*\]?\s*حديث (?:حسن|صحيح|ضعيف)|متفق عليه", matn_raw)[0]
    return narrator, clean_text(matn_raw), source


bounds = []
fahras_at = next((k for k, (pi, li, ln) in enumerate(flat) if ln.strip() == "الفهرس"), len(flat))
for k, (pi, li) in enumerate(an_lines):
    if k + 1 < len(an_lines):
        npi, nli = an_lines[k + 1]
    else:
        npi, nli = flat[fahras_at][0], flat[fahras_at][1]
    span_lines, span_pages = [], []
    for q in range(pi, npi + 1):
        lns = pages[q]["text"].split("\n")
        a = li if q == pi else 0
        b = nli if q == npi else len(lns)
        if a < b:
            span_lines.extend(lns[a:b])
            span_pages.append(pages[q]["page"])
    bounds.append((span_lines, span_pages))

all_hadiths = []
for (number, title), (span_lines, span_pages) in zip(hadith_titles, bounds):
    narrator, text, source = parse_span(span_lines)
    assert narrator and text and source, f"empty field in hadith {number}"
    all_hadiths.append({
        "hadith_number": number,
        "title": title,
        "narrator": narrator,
        "text": text,
        "source": source,
        "pages": span_pages,
    })
    print(f"{number}/42 pages={span_pages} :: {title}")

assert len(all_hadiths) == 42
assert all_hadiths[0]["pages"] == [3]
assert all_hadiths[1]["pages"] == [4, 5]
assert all_hadiths[2]["pages"] == [5]
assert all_hadiths[0]["text"].startswith("إنما الأعمال بالنيات")
assert all_hadiths[1]["text"].startswith("بينما")
assert all_hadiths[17]["narrator"] == "أبي ذر جندب بن جنادة وأبي عبد الرحمن معاذ بن جبل رضي الله تعالى عنهما"

with open("../data/hadith_extraction.json", "w", encoding="utf-8") as f:
    json.dump(all_hadiths, f, ensure_ascii=False, indent=2)

print("Saved to ../data/hadith_extraction.json")
```

```python
import json

input_path = "../data/hadith_extraction_clean.json"

with open(input_path, "r", encoding="utf-8") as f:
    hadiths = json.load(f)

print("Number of hadiths:", len(hadiths))
```

```python
required_fields = [
    "hadith_number",
    "id_label",
    "title",
    "narrator",
    "hadith_text",
    "source",
    "sharh",
    "rawi_bio",
    "pages",
    "page_title",
]

for hadith in hadiths:
    missing = [
        field
        for field in required_fields
        if field not in hadith
    ]

    if missing:
        print(
            f"Hadith {hadith.get('hadith_number')} "
            f"missing: {missing}"
        )
```

```python
seen = set()

for hadith in hadiths:
    number = hadith["hadith_number"]

    if number in seen:
        print(f"❌ Duplicate hadith: {number}")

    seen.add(number)
```
